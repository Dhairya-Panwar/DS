```python
# ==============================================================================
# 1. Python Errors and Built-in Exceptions
# ==============================================================================
# These lines are commented out because they would crash the script.
# Uncomment them one by one to see the error messages.

# print(undefined_variable)  # NameError
# result = "hello" + 5       # TypeError
# my_list = [1, 2, 3]
# print(my_list[5])          # IndexError
# number = int("abc")        # ValueError
# my_dict = {"a": 1}
# print(my_dict["b"])        # KeyError
# result = 10 / 0            # ZeroDivisionError

print("Common exceptions include NameError, TypeError, IndexError, ValueError, etc.")


# ==============================================================================
# 2. Basic Exception Handling: try...except
# ==============================================================================
print("try...except Block")
try:
    user_input = input("Enter a number: ")
    number = int(user_input)
    result = 100 / number
    print(f"100 divided by your number is: {result}")
except:
    print("An error occurred! Please enter a valid non-zero number.")


# ==============================================================================
# 3. Catching Specific Exceptions
# ==============================================================================
print("Catching Specific Exceptions")
try:
    user_input = input("Enter another number: ")
    number = int(user_input)
    result = 100 / number
    print(f"100 divided by your number is: {result}")
except ZeroDivisionError:
    print("An error occurred! Please enter a valid non-zero number.")
except ValueError:
    print("Error: That was not a valid number. Please enter digits only.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")


# ==============================================================================
# 4. The 'finally' Block
# ==============================================================================
print("The 'finally' Block")
f = None
try:
    print("Attempting to open and write to a file...")
    f = open("test_file.txt", "w")
    f.write("Hello, world!")
    # To demonstrate, let's cause an error after writing:
    result = 10 / 0
except ZeroDivisionError:
    print("Caught a division by zero error!")
finally:
    print("Executing the finally block...")
    if f:
        f.close()
        print("File has been closed.")


# ==============================================================================
# 5. The 'else' Block in a try...except statement
# ==============================================================================
print("The else Block")
try:
    user_input = input("Enter a number to find its reciprocal: ")
    number = float(user_input)
    reciprocal = 1 / number
except ValueError:
    print("Invalid input. Please enter a numeric value.")
except ZeroDivisionError:
    print("Cannot calculate the reciprocal of zero.")
else:
    print(f"The reciprocal of {number} is {reciprocal}.")
    print("Calculation was successful!")
finally:
    print("This concludes the reciprocal calculation attempt.")


# ==============================================================================
# 6. Raising an Exception
# ==============================================================================
print("Raising an Exception with raise")
def set_user_age(age):
    if not isinstance(age, int):
        raise TypeError("Age must be an integer.")
    if age < 0:
        raise ValueError("Age cannot be negative.")
    if age > 120:
        raise ValueError("Age is unrealistic. Please enter a valid age.")
    print(f"User age set to: {age}")

try:
    set_user_age(25)      # This will work
    set_user_age("twenty") # This will raise a TypeError
    set_user_age(-5)      # This will raise a ValueError
except ValueError as e:
    print(f"ValueError caught: {e}")
except TypeError as e:
    print(f"TypeError caught: {e}")

```
